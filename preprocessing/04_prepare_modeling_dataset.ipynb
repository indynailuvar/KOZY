Import library

```python
import pandas as pd
import numpy as np

from sklearn.preprocessing import StandardScaler

import os


pd.set_option(
    "display.max_columns",
    None
)
```

Load dataset clean

```python
df = pd.read_csv(
    "../analysis/kozy_dataset_clean.csv"
)


print(df.shape)

df.head()
```

Cek missing value sebelum treatment

```python
missing = pd.DataFrame({

    "jumlah_missing":
    df.isnull().sum(),

    "persentase":
    (df.isnull().sum()/len(df))*100

})


print(missing[missing["jumlah_missing"]>0])
```

Feature Selection

```python
drop_features = [

    "nama_kost",
    "sumber"

]


df = df.drop(
    columns=drop_features,
    errors="ignore"
)
```

Feature Engineering

Skor fasilitas

```python
fasilitas = [

"ac",
"wifi",
"kmandi_dalam",
"lemari",
"meja_belajar",
"kasur",
"tv",
"pmotor",
"pmobil"

]
```

```python
df["skor_fasilitas"] = (
    df[fasilitas]
    .sum(axis=1)
)
```

Popularity score

```python
df["popularitas_score"] = (
    df["jumlah_dilihat"].fillna(0)
    +
    df["jumlah_ulasan"].fillna(0)
)
```

Missing Value Treatment

```python
numeric_features = df.select_dtypes(
    include=["int64","float64"]
).columns



for col in numeric_features:

    df[col] = df[col].fillna(
        df[col].median()
    )
```

```python
categorical_features = df.select_dtypes(
    include=["object"]
).columns



for col in categorical_features:

    df[col] = df[col].fillna(
        df[col].mode()[0]
    )
```

Encoding Categorical

```python
df = pd.get_dummies(
    df,
    columns=[
        "jenis",
        "lokasi"
    ],
    drop_first=True
)
```

Scaling Numerical

```python
scaler = StandardScaler()


numeric_features = df.select_dtypes(
    include=["int64","float64"]
).columns



df[numeric_features] = scaler.fit_transform(
    df[numeric_features]
)
```
